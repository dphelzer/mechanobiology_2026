```python
import scanpy as sc
import anndata as ad
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

>just fixing the data

```python
WT = ad.read_h5ad("WT_FAPs_cell.h5ad")
mdx = ad.read_h5ad("mdx_FAPs_cell.h5ad")
TG = ad.read_h5ad("TG_FAPs_cell.h5ad")
```

```python
all_FAPs = ad.concat([WT, mdx, TG])
```

```python
# get normalized count data from the "raw" slot of the anndata object
all_FAPs_raw = all_FAPs.raw.to_adata()
all_FAPs_raw.obsm = all_FAPs.obsm
all_FAPs = all_FAPs_raw
```

```python
all_FAPs.X.shape
```

>22234 genes instead of 2000 that was in the .X slot previously (from scaled data)

---
>extract the percentage of cells expressing each gene within group

```python
dotplot = sc.pl.dotplot(
    all_FAPs,
    all_FAPs.var_names,
    groupby='condition',
    return_fig=True,
    show=False
)
```

```python
print(dotplot.dot_size_df['COL1A1']) 
```

>these are the percentages of cells expressing COL1A1 in each condition, which is in line with the results from the DE analysis

```python
# get the percentage data out and transpose to long format
dot_data = dotplot.dot_size_df.T
print(dot_data.head())
```

```python
# save
dot_data.to_excel("differential expression FAPs/percent_genes_per_group_FAPs_cell.xlsx")
```
