```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
```

```python
GO_terms = pd.read_excel("myonuclei_DE/GO terms.xlsx", sheet_name=1)
print(GO_terms.head())
```

```python
# plot parameters
import matplotlib as mpl
mpl.rcParams['font.family'] = 'Arial'
mpl.rcParams['font.size'] = 8
mpl.rcParams['svg.fonttype'] = 'none'
```

```python
# use a continuous colormap for the p‑values
cmap = plt.cm.viridis
norm = mpl.colors.Normalize(vmin=GO_terms['pval'].min(),
                            vmax=GO_terms['pval'].max())

# map each p‑value to a color
bar_colors = cmap(norm(GO_terms['pval']))
```
